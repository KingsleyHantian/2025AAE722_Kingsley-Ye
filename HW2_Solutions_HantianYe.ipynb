# Chapter 6: Question 1 - Stepwise Selection Methods

## Part (a): Which model with k predictors has the smallest training RSS?

**Answer: Best subset selection**

**Explanation:**
- **Best subset selection** evaluates all possible combinations of k predictors and chooses the one with the smallest RSS
- **Forward/Backward stepwise** use greedy algorithms that may not find the globally optimal combination
- Best subset considers all combinations, guaranteeing the smallest training RSS

## Part (b): Which model with k predictors has the smallest test RSS?

**Answer: Cannot be determined a priori - depends on the specific dataset and bias-variance tradeoff**

**Explanation:**
- The model with smallest training RSS may not have smallest test RSS due to overfitting
- Forward/backward stepwise may have better generalization if they avoid overfitting
- Depends on true relationship, noise level, sample size, and number of predictors

## Part (c): True or False Questions

### i. Forward stepwise subset property
**Answer: True** - Forward stepwise adds predictors sequentially, so k-variable model is always a subset of (k+1)-variable model

### ii. Backward stepwise subset property  
**Answer: True** - Backward stepwise removes predictors sequentially, so k-variable model is always a subset of (k+1)-variable model

### iii. Backward ⊆ Forward
**Answer: False** - Different selection paths can lead to completely different predictor sets

### iv. Forward ⊆ Backward
**Answer: False** - Same reasoning as (iii)

### v. Best subset ⊆ Best subset
**Answer: False** - Best subset independently optimizes for each k, no guarantee of subset relationship

# Chapter 6: Question 2 - Regularization Methods

## Part (a): The lasso, relative to least squares, is:

**Answer: iii. Less flexible and hence will give improved prediction accuracy when its increase in bias is less than its decrease in variance.**

**Explanation:**
- **Lasso** adds L1 penalty: λ∑|βⱼ|, performing regularization and variable selection
- **Less flexible** than least squares due to constraints
- **Bias-Variance Tradeoff:** Increases bias, decreases variance
- **Improves prediction** when bias increase < variance decrease

## Part (b): Ridge regression, relative to least squares, is:

**Answer: iii. Less flexible and hence will give improved prediction accuracy when its increase in bias is less than its decrease in variance.**

**Explanation:**
- **Ridge** adds L2 penalty: λ∑βⱼ², shrinking coefficients but not eliminating them
- **Less flexible** than least squares due to constraints
- **Same bias-variance principle** as lasso
- **Key difference:** Ridge shrinks all coefficients, lasso can eliminate variables

## Part (c): Non-linear methods, relative to least squares, are:

**Answer: i. More flexible and hence will give improved prediction accuracy when its increase in bias is less than its decrease in variance.**

**Explanation:**
- **Non-linear methods** can capture complex relationships
- **More flexible** than least squares (which only fits linear relationships)
- **Bias-Variance Tradeoff:** Can decrease bias, can increase variance
- **Improves prediction** when bias decrease > variance increase
- **Note:** The "increase in bias" is actually a decrease (negative increase)

## Summary Table

| Method | Flexibility vs LS | Condition for Improvement |
|--------|------------------|---------------------------|
| **Lasso** | Less flexible | Bias increase < Variance decrease |
| **Ridge** | Less flexible | Bias increase < Variance decrease |
| **Non-linear** | More flexible | Bias decrease > Variance increase |

## Question 14 Parts (c)-(h): Classification Methods Implementation


```python
# Part (c): Split Data into Training and Test Sets
features = ['weight', 'displacement', 'horsepower', 'cylinders']
X = Auto[features]
y = Auto['mpg01']

# Split the data into training and test sets (70-30 split)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f"Training set size: {X_train.shape[0]}")
print(f"Test set size: {X_test.shape[0]}")
print(f"\nTraining set mpg01 distribution:")
print(y_train.value_counts(normalize=True))
print(f"\nTest set mpg01 distribution:")
print(y_test.value_counts(normalize=True))
```

```python
# Parts (d)-(h): Implement All Classification Methods

# Initialize results storage
results = {}

# Part (d): Linear Discriminant Analysis (LDA)
print("=" * 50)
print("PART (d): LINEAR DISCRIMINANT ANALYSIS (LDA)")
print("=" * 50)
lda = LinearDiscriminantAnalysis()
lda.fit(X_train, y_train)
lda_pred = lda.predict(X_test)
lda_test_error = 1 - accuracy_score(y_test, lda_pred)
results['LDA'] = lda_test_error

print(f"LDA Test Error: {lda_test_error:.4f}")
print(f"LDA Test Accuracy: {accuracy_score(y_test, lda_pred):.4f}")

# Part (e): Quadratic Discriminant Analysis (QDA)
print("\n" + "=" * 50)
print("PART (e): QUADRATIC DISCRIMINANT ANALYSIS (QDA)")
print("=" * 50)
qda = QuadraticDiscriminantAnalysis()
qda.fit(X_train, y_train)
qda_pred = qda.predict(X_test)
qda_test_error = 1 - accuracy_score(y_test, qda_pred)
results['QDA'] = qda_test_error

print(f"QDA Test Error: {qda_test_error:.4f}")
print(f"QDA Test Accuracy: {accuracy_score(y_test, qda_pred):.4f}")

# Part (f): Logistic Regression
print("\n" + "=" * 50)
print("PART (f): LOGISTIC REGRESSION")
print("=" * 50)
logreg = LogisticRegression(random_state=42, max_iter=1000)
logreg.fit(X_train, y_train)
logreg_pred = logreg.predict(X_test)
logreg_test_error = 1 - accuracy_score(y_test, logreg_pred)
results['Logistic Regression'] = logreg_test_error

print(f"Logistic Regression Test Error: {logreg_test_error:.4f}")
print(f"Logistic Regression Test Accuracy: {accuracy_score(y_test, logreg_pred):.4f}")

# Display coefficients
print("\nCoefficients:")
for feature, coef in zip(features, logreg.coef_[0]):
    print(f"  {feature}: {coef:.4f}")
print(f"  Intercept: {logreg.intercept_[0]:.4f}")

# Part (g): Naive Bayes
print("\n" + "=" * 50)
print("PART (g): NAIVE BAYES")
print("=" * 50)
nb = GaussianNB()
nb.fit(X_train, y_train)
nb_pred = nb.predict(X_test)
nb_test_error = 1 - accuracy_score(y_test, nb_pred)
results['Naive Bayes'] = nb_test_error

print(f"Naive Bayes Test Error: {nb_test_error:.4f}")
print(f"Naive Bayes Test Accuracy: {accuracy_score(y_test, nb_pred):.4f}")
```

```python
# Part (h): K-Nearest Neighbors (KNN)
print("\n" + "=" * 50)
print("PART (h): K-NEAREST NEIGHBORS (KNN)")
print("=" * 50)

# Test different values of K
k_values = [1, 3, 5, 7, 9, 11, 13, 15]
knn_errors = []

print("KNN Results for Different K Values:")
print("K\tTest Error\tTest Accuracy")
print("-" * 30)

for k in k_values:
    # Fit KNN model
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    
    # Make predictions
    knn_pred = knn.predict(X_test)
    
    # Calculate test error and accuracy
    test_error = 1 - accuracy_score(y_test, knn_pred)
    test_accuracy = accuracy_score(y_test, knn_pred)
    
    knn_errors.append(test_error)
    
    print(f"{k}\t{test_error:.4f}\t\t{test_accuracy:.4f}")

# Find the best K
best_k_idx = np.argmin(knn_errors)
best_k = k_values[best_k_idx]
best_error = knn_errors[best_k_idx]
results[f'KNN (K={best_k})'] = best_error

print(f"\nBest K: {best_k} with test error: {best_error:.4f}")

# Plot KNN results
plt.figure(figsize=(10, 6))
plt.plot(k_values, knn_errors, 'bo-', linewidth=2, markersize=8)
plt.xlabel('K (Number of Neighbors)')
plt.ylabel('Test Error')
plt.title('KNN Test Error vs K')
plt.grid(True, alpha=0.3)
plt.xticks(k_values)

# Highlight the best K
plt.plot(best_k, best_error, 'ro', markersize=12, label=f'Best K={best_k}')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Question 14 Summary of Results
print("\n" + "=" * 60)
print("QUESTION 14: SUMMARY OF ALL METHODS")
print("=" * 60)

# Create summary table
results_summary = pd.DataFrame({
    'Method': list(results.keys()),
    'Test Error': list(results.values()),
    'Test Accuracy': [1-error for error in results.values()]
})

print(results_summary.to_string(index=False, float_format='%.4f'))

# Find the best method
best_method_idx = results_summary['Test Error'].idxmin()
best_method = results_summary.loc[best_method_idx, 'Method']
best_method_error = results_summary.loc[best_method_idx, 'Test Error']

print(f"\n🏆 Best Method: {best_method} with test error: {best_method_error:.4f}")

# Plot comparison
plt.figure(figsize=(12, 6))
methods = results_summary['Method']
errors = results_summary['Test Error']

bars = plt.bar(methods, errors, color=['skyblue', 'lightgreen', 'lightcoral', 'lightyellow', 'lightpink'])
plt.xlabel('Classification Method')
plt.ylabel('Test Error')
plt.title('Comparison of Test Errors Across Different Methods')
plt.xticks(rotation=45, ha='right')
plt.grid(True, alpha=0.3, axis='y')

# Add value labels on bars
for bar, error in zip(bars, errors):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.005, 
             f'{error:.4f}', ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

# Question 8: Cross-Validation on Simulated Data

We will now perform cross-validation on a simulated data set to compare different polynomial models.


```python
# Question 8 Part (a): Generate Simulated Data
np.random.seed(1)

# Generate simulated data
n = 100  # number of observations
X = np.random.uniform(0, 1, n)  # X ~ Uniform(0,1)
epsilon = np.random.normal(0, 0.1, n)  # epsilon ~ N(0, 0.1^2)
Y = 2 + 3*X + 4*X**2 + epsilon  # Y = 2 + 3X + 4X^2 + epsilon

# Create DataFrame
data = pd.DataFrame({'X': X, 'Y': Y})

print(f"n = {n} (number of observations)")
print(f"p = 1 (number of predictors - just X)")
print(f"\nModel used to generate the data:")
print("Y = β₀ + β₁X + β₂X² + ε")
print("where β₀ = 2, β₁ = 3, β₂ = 4, and ε ~ N(0, 0.1²)")

print(f"\nData summary:")
print(data.describe())
```

```python
# Question 8 Part (b): Create Scatterplot
plt.figure(figsize=(10, 6))
plt.scatter(X, Y, alpha=0.6, s=50)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatterplot of X against Y')
plt.grid(True, alpha=0.3)

# Add the true relationship curve
X_true = np.linspace(0, 1, 100)
Y_true = 2 + 3*X_true + 4*X_true**2
plt.plot(X_true, Y_true, 'r-', linewidth=2, label='True relationship: Y = 2 + 3X + 4X²')
plt.legend()
plt.tight_layout()
plt.show()

print("Comments on the scatterplot:")
print("- The data shows a clear quadratic relationship between X and Y")
print("- The points follow a curved pattern, not linear")
print("- There is some noise around the true quadratic curve due to the error term")
print("- The relationship is monotonic and increasing over the range [0,1]")
```

```python
# Question 8 Parts (c)-(d): LOOCV Analysis
print("=" * 60)
print("QUESTION 8: LOOCV ANALYSIS")
print("=" * 60)

# Prepare data
X_data = data[['X']].values
y_data = data['Y'].values

# Define models to test
models = {
    'Linear': 1,
    'Quadratic': 2, 
    'Cubic': 3,
    'Quartic': 4
}

# Perform LOOCV for each model
loocv_errors = {}

print("LOOCV Results (Random Seed = 1):")
print("Model\t\tLOOCV Error")
print("-" * 30)

for model_name, degree in models.items():
    # Create polynomial features
    poly_features = PolynomialFeatures(degree=degree, include_bias=True)
    X_poly = poly_features.fit_transform(X_data)
    
    # Perform LOOCV
    loo = LeaveOneOut()
    errors = []
    
    for train_idx, val_idx in loo.split(X_poly):
        X_train, X_val = X_poly[train_idx], X_poly[val_idx]
        y_train, y_val = y_data[train_idx], y_data[val_idx]
        
        # Fit model
        model = LinearRegression()
        model.fit(X_train, y_train)
        
        # Predict and calculate error
        y_pred = model.predict(X_val)
        error = (y_val - y_pred) ** 2
        errors.append(error)
    
    # Calculate mean LOOCV error
    loocv_error = np.mean(errors)
    loocv_errors[model_name] = loocv_error
    
    print(f"{model_name}\t\t{loocv_error:.6f}")

# Find the best model
best_model = min(loocv_errors, key=loocv_errors.get)
best_error = loocv_errors[best_model]

print(f"\n🏆 Best model: {best_model} with LOOCV error: {best_error:.6f}")

# Test with different random seed (Part d)
print(f"\n" + "=" * 40)
print("PART (d): DIFFERENT RANDOM SEED")
print("=" * 40)

np.random.seed(42)
X_new = np.random.uniform(0, 1, n)
epsilon_new = np.random.normal(0, 0.1, n)
Y_new = 2 + 3*X_new + 4*X_new**2 + epsilon_new

X_data_new = X_new.reshape(-1, 1)
y_data_new = Y_new

loocv_errors_new = {}
print("LOOCV Results (Random Seed = 42):")
print("Model\t\tLOOCV Error")
print("-" * 30)

for model_name, degree in models.items():
    poly_features = PolynomialFeatures(degree=degree, include_bias=True)
    X_poly = poly_features.fit_transform(X_data_new)
    
    loo = LeaveOneOut()
    errors = []
    
    for train_idx, val_idx in loo.split(X_poly):
        X_train, X_val = X_poly[train_idx], X_poly[val_idx]
        y_train, y_val = y_data_new[train_idx], y_data_new[val_idx]
        
        model = LinearRegression()
        model.fit(X_train, y_train)
        y_pred = model.predict(X_val)
        error = (y_val - y_pred) ** 2
        errors.append(error)
    
    loocv_error = np.mean(errors)
    loocv_errors_new[model_name] = loocv_error
    print(f"{model_name}\t\t{loocv_error:.6f}")

print(f"\nAre the results the same? {loocv_errors == loocv_errors_new}")
print("Why? The results are different because we generated different data with different random seeds.")
print("The LOOCV procedure itself is deterministic, but the underlying data is different.")
```

```python
# Question 8 Parts (e)-(f): Analysis and Statistical Significance
print("\n" + "=" * 60)
print("QUESTION 8: ANALYSIS AND STATISTICAL SIGNIFICANCE")
print("=" * 60)

print("Part (e): Analysis of Best Model")
print(f"Best model: {best_model}")
print(f"LOOCV error: {best_error:.6f}")

print(f"\nIs this what we expected?")
print("Yes, this is expected because:")
print("1. The true data generating process is: Y = 2 + 3X + 4X² + ε")
print("2. This is a quadratic relationship (degree 2)")
print("3. The quadratic model should have the lowest LOOCV error")
print("4. Higher degree models (cubic, quartic) may overfit and have higher error")
print("5. Lower degree models (linear) may underfit and have higher error")

print(f"\n" + "=" * 50)
print("Part (f): Statistical Significance of Coefficients")
print("=" * 50)

# Fit each model and analyze coefficients
for model_name, degree in models.items():
    print(f"\n{model_name} Model (Y = β₀ + β₁X + ... + β_{degree}X^{degree} + ε):")
    print("-" * 50)
    
    # Create polynomial features
    poly_features = PolynomialFeatures(degree=degree, include_bias=True)
    X_poly = poly_features.fit_transform(X_data)
    
    # Fit model using statsmodels for statistical tests
    model_sm = sm.OLS(y_data, X_poly).fit()
    
    # Extract p-values for coefficients
    p_values = model_sm.pvalues
    significant_coeffs = p_values[p_values < 0.05]
    
    print(f"Significant coefficients (p < 0.05): {len(significant_coeffs)} out of {len(p_values)}")
    for i, (coeff_name, p_val) in enumerate(zip(model_sm.params.index, p_values)):
        significance = "***" if p_val < 0.001 else "**" if p_val < 0.01 else "*" if p_val < 0.05 else ""
        print(f"  {coeff_name}: p = {p_val:.6f} {significance}")

print(f"\n" + "="*60)
print("CONCLUSIONS:")
print("="*60)
print("1. The cross-validation results and statistical significance tests should agree:")
print("   - Models with significant coefficients should generally perform better in CV")
print("   - The quadratic model should show significant X² coefficient")
print("   - Higher-order terms in cubic/quartic models may not be significant")
print("\n2. Expected findings:")
print("   - Linear model: Only β₀ and β₁ should be significant")
print("   - Quadratic model: β₀, β₁, and β₂ should all be significant")
print("   - Cubic model: β₃ may not be significant (overfitting)")
print("   - Quartic model: β₃ and β₄ may not be significant (overfitting)")
print("\n3. This supports the cross-validation conclusion that quadratic is optimal")
```
