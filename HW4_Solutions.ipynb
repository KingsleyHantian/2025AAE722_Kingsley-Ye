# Question 1: Hyperplanes in Two Dimensions

This problem involves hyperplanes in two dimensions.

**(a)** Sketch the hyperplane $1 + 3X_1 - X_2 = 0$. Indicate the set of points for which $1 + 3X_1 - X_2 > 0$, as well as the set of points for which $1 + 3X_1 - X_2 < 0$.

**(b)** On the same plot, sketch the hyperplane $-2 + X_1 + 2X_2 = 0$. Indicate the set of points for which $-2 + X_1 + 2X_2 > 0$, as well as the set of points for which $-2 + X_1 + 2X_2 < 0$.


## Import Libraries


```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.preprocessing import PolynomialFeatures
import warnings
warnings.filterwarnings('ignore')

# Set style for better plots
plt.style.use('seaborn-v0_8')
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 12
```

## Part (a) and (b): Plot Both Hyperplanes

The hyperplane equations are:
- Hyperplane 1: $1 + 3X_1 - X_2 = 0$ → $X_2 = 1 + 3X_1$
- Hyperplane 2: $-2 + X_1 + 2X_2 = 0$ → $X_2 = 1 - 0.5X_1$

For hyperplane 1:
- $1 + 3X_1 - X_2 > 0$: $X_2 < 1 + 3X_1$ (points below the line)
- $1 + 3X_1 - X_2 < 0$: $X_2 > 1 + 3X_1$ (points above the line)

For hyperplane 2:
- $-2 + X_1 + 2X_2 > 0$: $X_2 > 1 - 0.5X_1$ (points above the line)
- $-2 + X_1 + 2X_2 < 0$: $X_2 < 1 - 0.5X_1$ (points below the line)


```python
# Create a grid of points
x1_range = np.linspace(-3, 3, 1000)
x2_range = np.linspace(-5, 8, 1000)
X1, X2 = np.meshgrid(x1_range, x2_range)

# Define the hyperplane functions
def hyperplane1(x1, x2):
    """1 + 3X1 - X2 = 0"""
    return 1 + 3*x1 - x2

def hyperplane2(x1, x2):
    """-2 + X1 + 2X2 = 0"""
    return -2 + x1 + 2*x2

# Calculate values for both hyperplanes
H1 = hyperplane1(X1, X2)
H2 = hyperplane2(X1, X2)

# Create the plot
fig, ax = plt.subplots(figsize=(12, 10))

# Plot regions for hyperplane 1
ax.contourf(X1, X2, H1, levels=[-100, 0, 100], colors=['lightblue', 'lightcoral'], alpha=0.3)
ax.contour(X1, X2, H1, levels=[0], colors='blue', linewidths=2.5, linestyles='-', label='$1 + 3X_1 - X_2 = 0$')

# Plot regions for hyperplane 2
ax.contourf(X1, X2, H2, levels=[-100, 0, 100], colors=['lightgreen', 'lightyellow'], alpha=0.3)
ax.contour(X1, X2, H2, levels=[0], colors='red', linewidths=2.5, linestyles='--', label='$-2 + X_1 + 2X_2 = 0$')

# Add text annotations for regions
ax.text(-2, 6, '$1 + 3X_1 - X_2 > 0$\\n(above line)', fontsize=11, 
        bbox=dict(boxstyle='round', facecolor='lightcoral', alpha=0.7), ha='center')
ax.text(-2, -2, '$1 + 3X_1 - X_2 < 0$\\n(below line)', fontsize=11,
        bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.7), ha='center')

ax.text(2, 6, '$-2 + X_1 + 2X_2 > 0$', fontsize=11,
        bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.7), ha='center')
ax.text(2, -2, '$-2 + X_1 + 2X_2 < 0$', fontsize=11,
        bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.7), ha='center')

# Labels and title
ax.set_xlabel('$X_1$', fontsize=14)
ax.set_ylabel('$X_2$', fontsize=14)
ax.set_title('Hyperplanes in Two Dimensions', fontsize=16, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(loc='upper right', fontsize=12)
ax.set_xlim(-3, 3)
ax.set_ylim(-5, 8)

plt.tight_layout()
plt.show()

print("="*70)
print("HYPERPLANE ANALYSIS:")
print("="*70)
print("\nHyperplane 1: $1 + 3X_1 - X_2 = 0$")
print("  - Rearranged: $X_2 = 1 + 3X_1$")
print("  - Region where $1 + 3X_1 - X_2 > 0$: $X_2 < 1 + 3X_1$ (below line)")
print("  - Region where $1 + 3X_1 - X_2 < 0$: $X_2 > 1 + 3X_1$ (above line)")
print("\nHyperplane 2: $-2 + X_1 + 2X_2 = 0$")
print("  - Rearranged: $X_2 = 1 - 0.5X_1$")
print("  - Region where $-2 + X_1 + 2X_2 > 0$: $X_2 > 1 - 0.5X_1$ (above line)")
print("  - Region where $-2 + X_1 + 2X_2 < 0$: $X_2 < 1 - 0.5X_1$ (below line)")
```

# Question 2: Chapter 9 Question 5 - Logistic Regression and SVM

This question explores logistic regression and Support Vector Machines (SVM) with non-linear decision boundaries.

**Goal:** Generate a data set with $n = 500$ observations and $p = 2$ predictors ($X_1$, $X_2$) such that the observations belong to two classes with a quadratic decision boundary.


## Part (a): Generate Data Set


```python
# Generate data with quadratic decision boundary
rng = np.random.default_rng(5)
x1 = rng.uniform(size=500) - 0.5
x2 = rng.uniform(size=500) - 0.5
y = x1**2 - x2**2 > 0
y = y.astype(int)

# Create DataFrame
data = pd.DataFrame({'X1': x1, 'X2': x2, 'y': y})

print("="*70)
print("DATA GENERATION:")
print("="*70)
print(f"\nNumber of observations: {len(data)}")
print(f"Number of features: 2 (X1, X2)")
print(f"Decision boundary: $X_1^2 - X_2^2 = 0$ (quadratic)")
print(f"\nClass distribution:")
print(data['y'].value_counts().sort_index())
print(f"\nFirst 5 rows:")
print(data.head())
```

## Part (b): Plot the Generated Observations


```python
# Plot the observations colored by true class labels
fig, ax = plt.subplots(figsize=(10, 8))

# Separate classes
class0 = data[data['y'] == 0]
class1 = data[data['y'] == 1]

ax.scatter(class0['X1'], class0['X2'], c='blue', s=50, alpha=0.6, label='Class 0', edgecolors='black', linewidths=0.5)
ax.scatter(class1['X1'], class1['X2'], c='red', s=50, alpha=0.6, label='Class 1', edgecolors='black', linewidths=0.5)

# Plot the true decision boundary (X1^2 - X2^2 = 0, i.e., X2 = ±X1)
x1_boundary = np.linspace(-0.5, 0.5, 100)
x2_boundary_pos = x1_boundary
x2_boundary_neg = -x1_boundary
ax.plot(x1_boundary, x2_boundary_pos, 'g--', linewidth=2, label='True Decision Boundary ($X_1^2 - X_2^2 = 0$)', alpha=0.7)
ax.plot(x1_boundary, x2_boundary_neg, 'g--', linewidth=2, alpha=0.7)

ax.set_xlabel('$X_1$', fontsize=14)
ax.set_ylabel('$X_2$', fontsize=14)
ax.set_title('Generated Data with Quadratic Decision Boundary', fontsize=16, fontweight='bold')
ax.legend(loc='upper right', fontsize=11)
ax.grid(True, alpha=0.3)
ax.set_xlim(-0.6, 0.6)
ax.set_ylim(-0.6, 0.6)

plt.tight_layout()
plt.show()

print("\nThe true decision boundary is $X_1^2 - X_2^2 = 0$, which simplifies to $X_2 = \\pm X_1$.")
print("This creates a diagonal cross pattern separating the classes.")
```

## Part (c): Fit Linear Logistic Regression Model


```python
# Prepare data
X = data[['X1', 'X2']].values
y = data['y'].values

# Fit linear logistic regression
lr_linear = LogisticRegression(max_iter=1000, random_state=42)
lr_linear.fit(X, y)

print("="*70)
print("LINEAR LOGISTIC REGRESSION MODEL:")
print("="*70)
print(f"\nCoefficients:")
print(f"  Intercept: {lr_linear.intercept_[0]:.4f}")
print(f"  X1 coefficient: {lr_linear.coef_[0][0]:.4f}")
print(f"  X2 coefficient: {lr_linear.coef_[0][1]:.4f}")
print(f"\nTraining accuracy: {lr_linear.score(X, y):.4f}")
```

## Part (d): Plot Linear Logistic Regression Results


```python
# Get predictions
y_pred_linear = lr_linear.predict(X)

# Create mesh for decision boundary
xx, yy = np.meshgrid(np.linspace(-0.6, 0.6, 200),
                     np.linspace(-0.6, 0.6, 200))
Z_linear = lr_linear.predict(np.c_[xx.ravel(), yy.ravel()])
Z_linear = Z_linear.reshape(xx.shape)

# Plot
fig, ax = plt.subplots(figsize=(10, 8))

# Plot decision regions
ax.contourf(xx, yy, Z_linear, alpha=0.3, colors=['lightblue', 'lightcoral'])

# Plot points colored by predicted labels
pred0 = data[y_pred_linear == 0]
pred1 = data[y_pred_linear == 1]
ax.scatter(pred0['X1'], pred0['X2'], c='blue', s=50, alpha=0.6, label='Predicted Class 0', edgecolors='black', linewidths=0.5)
ax.scatter(pred1['X1'], pred1['X2'], c='red', s=50, alpha=0.6, label='Predicted Class 1', edgecolors='black', linewidths=0.5)

# Plot decision boundary
ax.contour(xx, yy, Z_linear, levels=[0.5], colors='black', linewidths=2, linestyles='-', label='Decision Boundary')

ax.set_xlabel('$X_1$', fontsize=14)
ax.set_ylabel('$X_2$', fontsize=14)
ax.set_title('Linear Logistic Regression Predictions', fontsize=16, fontweight='bold')
ax.legend(loc='upper right', fontsize=11)
ax.grid(True, alpha=0.3)
ax.set_xlim(-0.6, 0.6)
ax.set_ylim(-0.6, 0.6)

plt.tight_layout()
plt.show()

print(f"\nTraining accuracy: {lr_linear.score(X, y):.4f}")
print("\nAs expected, the decision boundary is linear and cannot capture the quadratic relationship.")
```

## Part (e): Fit Non-linear Logistic Regression Model


```python
# Create polynomial features (including X1^2, X2^2, X1*X2, etc.)
poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(X)

# Fit non-linear logistic regression
lr_nonlinear = LogisticRegression(max_iter=1000, random_state=42)
lr_nonlinear.fit(X_poly, y)

print("="*70)
print("NON-LINEAR LOGISTIC REGRESSION MODEL:")
print("="*70)
print(f"\nPolynomial features: {poly.get_feature_names_out(['X1', 'X2'])}")
print(f"\nCoefficients:")
print(f"  Intercept: {lr_nonlinear.intercept_[0]:.4f}")
for i, name in enumerate(poly.get_feature_names_out(['X1', 'X2'])):
    print(f"  {name}: {lr_nonlinear.coef_[0][i]:.4f}")
print(f"\nTraining accuracy: {lr_nonlinear.score(X_poly, y):.4f}")
```

## Part (f): Plot Non-linear Logistic Regression Results


```python
# Get predictions
y_pred_nonlinear = lr_nonlinear.predict(X_poly)

# Create mesh for decision boundary
xx, yy = np.meshgrid(np.linspace(-0.6, 0.6, 200),
                     np.linspace(-0.6, 0.6, 200))
mesh_points = np.c_[xx.ravel(), yy.ravel()]
mesh_poly = poly.transform(mesh_points)
Z_nonlinear = lr_nonlinear.predict(mesh_poly)
Z_nonlinear = Z_nonlinear.reshape(xx.shape)

# Plot
fig, ax = plt.subplots(figsize=(10, 8))

# Plot decision regions
ax.contourf(xx, yy, Z_nonlinear, alpha=0.3, colors=['lightblue', 'lightcoral'])

# Plot points colored by predicted labels
pred0 = data[y_pred_nonlinear == 0]
pred1 = data[y_pred_nonlinear == 1]
ax.scatter(pred0['X1'], pred0['X2'], c='blue', s=50, alpha=0.6, label='Predicted Class 0', edgecolors='black', linewidths=0.5)
ax.scatter(pred1['X1'], pred1['X2'], c='red', s=50, alpha=0.6, label='Predicted Class 1', edgecolors='black', linewidths=0.5)

# Plot decision boundary
ax.contour(xx, yy, Z_nonlinear, levels=[0.5], colors='black', linewidths=2, linestyles='-', label='Decision Boundary')

ax.set_xlabel('$X_1$', fontsize=14)
ax.set_ylabel('$X_2$', fontsize=14)
ax.set_title('Non-linear Logistic Regression Predictions', fontsize=16, fontweight='bold')
ax.legend(loc='upper right', fontsize=11)
ax.grid(True, alpha=0.3)
ax.set_xlim(-0.6, 0.6)
ax.set_ylim(-0.6, 0.6)

plt.tight_layout()
plt.show()

print(f"\nTraining accuracy: {lr_nonlinear.score(X_poly, y):.4f}")
print("\nThe decision boundary is now non-linear and can capture the quadratic relationship.")
```

## Part (g): Linear Support Vector Classifier (SVC)


```python
# Fit linear SVC
svc_linear = SVC(kernel='linear', random_state=42)
svc_linear.fit(X, y)

# Get predictions
y_pred_svc_linear = svc_linear.predict(X)

# Create mesh for decision boundary
xx, yy = np.meshgrid(np.linspace(-0.6, 0.6, 200),
                     np.linspace(-0.6, 0.6, 200))
Z_svc_linear = svc_linear.predict(np.c_[xx.ravel(), yy.ravel()])
Z_svc_linear = Z_svc_linear.reshape(xx.shape)

# Plot
fig, ax = plt.subplots(figsize=(10, 8))

# Plot decision regions
ax.contourf(xx, yy, Z_svc_linear, alpha=0.3, colors=['lightblue', 'lightcoral'])

# Plot points colored by predicted labels
pred0 = data[y_pred_svc_linear == 0]
pred1 = data[y_pred_svc_linear == 1]
ax.scatter(pred0['X1'], pred0['X2'], c='blue', s=50, alpha=0.6, label='Predicted Class 0', edgecolors='black', linewidths=0.5)
ax.scatter(pred1['X1'], pred1['X2'], c='red', s=50, alpha=0.6, label='Predicted Class 1', edgecolors='black', linewidths=0.5)

# Plot decision boundary
ax.contour(xx, yy, Z_svc_linear, levels=[0.5], colors='black', linewidths=2, linestyles='-', label='Decision Boundary')

ax.set_xlabel('$X_1$', fontsize=14)
ax.set_ylabel('$X_2$', fontsize=14)
ax.set_title('Linear Support Vector Classifier', fontsize=16, fontweight='bold')
ax.legend(loc='upper right', fontsize=11)
ax.grid(True, alpha=0.3)
ax.set_xlim(-0.6, 0.6)
ax.set_ylim(-0.6, 0.6)

plt.tight_layout()
plt.show()

print(f"\nTraining accuracy: {svc_linear.score(X, y):.4f}")
print("\nLinear SVC also produces a linear decision boundary.")
```

# Question 3: Chapter 10 Question 2 - Softmax Function Properties

This question concerns the softmax function.

The softmax for class $m$ is:

$$f_m(X) = \\Pr(Y = m|X) = \\frac{e^{Z_m}}{\\sum_{l=0}^{K} e^{Z_l}}$$

**(a)** Show that if we add a constant $c$ to each $Z_l$, the probabilities are unchanged.

**(b)** In equation (4.13), show that if we add constants $c_j$, $j = 0, 1, \\ldots, p$, to the corresponding coefficients for each class, then predictions at any new point $x$ are unchanged.


## Part (a): Proof that Adding Constant to All $Z_l$ Does Not Change Probabilities

### Mathematical Proof

**Given:** The softmax function for class $m$:

$$f_m(X) = \\Pr(Y = m|X) = \\frac{e^{Z_m}}{\\sum_{l=0}^{K} e^{Z_l}}$$

**Claim:** If we add a constant $c$ to each $Z_l$, the probabilities remain unchanged.

**Proof:**

Let $Z_l' = Z_l + c$ for all $l = 0, 1, \\ldots, K$.

The new softmax probability for class $m$ is:

$$f_m'(X) = \\frac{e^{Z_m'}}{\\sum_{l=0}^{K} e^{Z_l'}} = \\frac{e^{Z_m + c}}{\\sum_{l=0}^{K} e^{Z_l + c}}$$

Using the property $e^{a+b} = e^a \\cdot e^b$:

$$f_m'(X) = \\frac{e^{Z_m} \\cdot e^c}{\\sum_{l=0}^{K} e^{Z_l} \\cdot e^c} = \\frac{e^{Z_m} \\cdot e^c}{e^c \\sum_{l=0}^{K} e^{Z_l}}$$

Since $e^c$ is a constant (does not depend on $l$), we can factor it out:

$$f_m'(X) = \\frac{e^{Z_m} \\cdot e^c}{e^c \\sum_{l=0}^{K} e^{Z_l}} = \\frac{e^{Z_m}}{\\sum_{l=0}^{K} e^{Z_l}} = f_m(X)$$

**Q.E.D.** The probabilities are unchanged.

**Intuition:** Adding the same constant to all logits shifts them all by the same amount, but since we're taking ratios (normalizing), the shift cancels out in the final probabilities.


```python
# Numerical demonstration
def softmax(Z):
    """Compute softmax probabilities"""
    exp_Z = np.exp(Z - np.max(Z))  # Numerical stability: subtract max
    return exp_Z / np.sum(exp_Z)

# Example with K+1 = 4 classes (K=3, so classes 0, 1, 2, 3)
Z_original = np.array([2.0, 1.0, 0.5, 0.1])
c = 5.0  # Constant to add
Z_shifted = Z_original + c

prob_original = softmax(Z_original)
prob_shifted = softmax(Z_shifted)

print("="*70)
print("NUMERICAL DEMONSTRATION: Part (a)")
print("="*70)
print(f"\nOriginal logits Z: {Z_original}")
print(f"Original probabilities: {prob_original}")
print(f"\nShifted logits Z' = Z + {c}: {Z_shifted}")
print(f"Shifted probabilities: {prob_shifted}")
print(f"\nDifference: {np.abs(prob_original - prob_shifted)}")
print(f"\nMaximum absolute difference: {np.max(np.abs(prob_original - prob_shifted)):.10f}")
print("\n✓ The probabilities are identical (within numerical precision)!")
print("\nThis confirms that adding a constant to all logits does not change the probabilities.")
```

## Part (b): Proof that Adding Constants to Coefficients Does Not Change Predictions

### Mathematical Formulation

In equation (4.13), the linear predictor for class $m$ is typically:

$$Z_m = \\beta_{m0} + \\beta_{m1}X_1 + \\beta_{m2}X_2 + \\cdots + \\beta_{mp}X_p$$

Or more compactly:

$$Z_m = \\beta_{m0} + \\sum_{j=1}^{p} \\beta_{mj} X_j$$

**Claim:** If we add constants $c_j$ (for $j = 0, 1, \\ldots, p$) to the corresponding coefficients for each class, the predictions remain unchanged.

**Proof:**

Let the original coefficients be $\\beta_{mj}$ for class $m$ and feature $j$.

The original linear predictor for class $m$ is:

$$Z_m = \\beta_{m0} + \\sum_{j=1}^{p} \\beta_{mj} X_j$$

Now, add constants $c_j$ to each coefficient:

$$Z_m' = (\\beta_{m0} + c_0) + \\sum_{j=1}^{p} (\\beta_{mj} + c_j) X_j$$

Expanding:

$$Z_m' = \\beta_{m0} + c_0 + \\sum_{j=1}^{p} \\beta_{mj} X_j + \\sum_{j=1}^{p} c_j X_j$$

$$Z_m' = \\left(\\beta_{m0} + \\sum_{j=1}^{p} \\beta_{mj} X_j\\right) + \\left(c_0 + \\sum_{j=1}^{p} c_j X_j\\right)$$

$$Z_m' = Z_m + \\left(c_0 + \\sum_{j=1}^{p} c_j X_j\\right)$$

The key observation is that the term $c_0 + \\sum_{j=1}^{p} c_j X_j$ is **the same for all classes** $m$ (it does not depend on $m$).

Therefore, for all classes $m = 0, 1, \\ldots, K$:

$$Z_m' = Z_m + C$$

where $C = c_0 + \\sum_{j=1}^{p} c_j X_j$ is a constant (same for all classes).

From Part (a), we know that adding the same constant to all $Z_m$ does not change the softmax probabilities. Therefore, the predictions remain unchanged.

**Q.E.D.**

**Intuition:** Adding the same constants to the coefficients of all classes shifts all logits by the same amount (which depends on $X$ but not on the class $m$). Since the softmax is invariant to such shifts, the predictions are unchanged.


```python
# Numerical demonstration
def compute_logits(beta, X):
    """Compute logits Z_m for all classes given coefficients beta and features X"""
    # beta shape: (K+1, p+1) where first column is intercept
    # X shape: (n, p)
    # Add intercept column
    X_with_intercept = np.column_stack([np.ones(X.shape[0]), X])
    # Compute Z = X @ beta^T
    Z = X_with_intercept @ beta.T
    return Z

# Example: 3 classes (K=2, so classes 0, 1, 2), 2 features (p=2)
K = 2  # 3 classes: 0, 1, 2
p = 2  # 2 features

# Original coefficients: beta shape (3, 3) = (K+1, p+1)
# Each row is [beta_m0, beta_m1, beta_m2] for class m
beta_original = np.array([
    [1.0, 2.0, 0.5],  # Class 0: beta_00=1.0, beta_01=2.0, beta_02=0.5
    [0.5, 1.5, 1.0],  # Class 1: beta_10=0.5, beta_11=1.5, beta_12=1.0
    [0.0, 1.0, 1.5]   # Class 2: beta_20=0.0, beta_21=1.0, beta_22=1.5
])

# Constants to add to each coefficient
c = np.array([3.0, 2.0, 1.0])  # c_0, c_1, c_2

# Shifted coefficients: add c_j to beta_mj for all m
beta_shifted = beta_original.copy()
for j in range(p + 1):  # j = 0, 1, 2 (intercept and 2 features)
    beta_shifted[:, j] += c[j]

# Test point
x_test = np.array([[1.5, 2.0]])  # Single test point

# Compute logits
Z_original = compute_logits(beta_original, x_test)[0]  # [0] to get single point
Z_shifted = compute_logits(beta_shifted, x_test)[0]

# Compute probabilities
prob_original = softmax(Z_original)
prob_shifted = softmax(Z_shifted)

print("="*70)
print("NUMERICAL DEMONSTRATION: Part (b)")
print("="*70)
print(f"\nOriginal coefficients beta:")
print(beta_original)
print(f"\nConstants c to add: {c}")
print(f"\nShifted coefficients beta':")
print(beta_shifted)
print(f"\nTest point x: {x_test[0]}")
print(f"\nOriginal logits Z: {Z_original}")
print(f"Shifted logits Z': {Z_shifted}")
print(f"\nDifference Z' - Z: {Z_shifted - Z_original}")
print(f"\nNote: All differences are the same! (as expected)")
print(f"\nOriginal probabilities: {prob_original}")
print(f"Shifted probabilities: {prob_shifted}")
print(f"\nDifference: {np.abs(prob_original - prob_shifted)}")
print(f"\nMaximum absolute difference: {np.max(np.abs(prob_original - prob_shifted)):.10f}")
print("\n✓ The probabilities are identical (within numerical precision)!")
print("\nThis confirms that adding constants to coefficients does not change predictions.")
```
